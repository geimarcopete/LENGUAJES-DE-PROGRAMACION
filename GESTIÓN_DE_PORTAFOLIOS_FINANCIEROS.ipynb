 **Gestión de Portafolios Financieros**

Eres parte del equipo de análisis financiero de una empresa que gestiona portafolios de inversión para varios clientes. Debes desarrollar un script en Python para ayudar a la empresa a gestionar información sobre las inversiones de los clientes y generar informes sobre el rendimiento de sus portafolios.

***Parte 1: Declaración de Variables y Listas***

**Datos de inversión inicial**

La empresa ha recopilado los siguientes datos sobre cuatro clientes


*   Clientes: Juan Perez, Maria Rodriguez, Carlos Lopez, Ana Gomez
*   Montos invertidos (en dólares): 10000, 15000, 20000, 18000
*   Rendimientos anuales(%): 5.5, 6.0, 4.0, 7.5





***Listas*:**

```python
clientes = ["Juan Perez", "Maria Rodriguez", "Carlos Lopez", "Ana Gomez"]
montos_invertidos = [10000, 15000, 20000, 18000]
rendimientos = [5.5, 6.0, 4.0, 7.5]
```

***Tuplas Para Almacenar la información*:**

---



```python
cliente_1 = (clientes[0], montos_invertidos[0], rendimientos[0])
cliente_2 = (clientes[1], montos_invertidos[1], rendimientos[1])
cliente_3 = (clientes[2], montos_invertidos[2], rendimientos[2])
cliente_4 = (clientes[3], montos_invertidos[3], rendimientos[3])

print("Parte 1 - Tuplas creadas:")
print(cliente_1)
print(cliente_2)
print(cliente_3)
print(cliente_4)
print("-"*50)
```

***Parte 2: Diccionarios y Condicionales***

***Almacenamiento de datos en un diccionario:***

```python
portafolio = {
    "Juan Perez": {"monto_invertido": 10000, "rendimiento": 5.5},
    "Maria Rodriguez": {"monto_invertido": 15000, "rendimiento": 6.0},
    "Carlos Lopez": {"monto_invertido": 20000, "rendimiento": 4.0},
    "Ana Gomez": {"monto_invertido": 18000, "rendimiento": 7.5}
}

print(portafolio)
```

***Evalución de rendimiento de los clientes:***

```python
print("Parte 2 - Clientes con rendimiento > 5%:")
for cliente, datos in portafolio.items():
    if datos["rendimiento"] > 5:
        print(f"{cliente} tiene un rendimiento de {datos['rendimiento']}%")
print("-"*50)
```

***Parte 3: Cálculos con Ciclos***

La empresa desea calcular el rendimiento total del portafolio, que se define como la suma de todos los rendimientos en dólares.

***Formula:***

```python
from IPython.display import display, Math

display(Math(r"\text{rendimiento total} = \frac{\text{monto invertido} \times \text{rendimiento}}{100}"))
```

*Rendimiento total:*

```python
# Calcular rendimiento total
rendimiento_total = 0
for cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento"]) / 100
    rendimiento_total += rendimiento_cliente

print(f"Parte 3 - El rendimiento total del portafolio es: ${rendimiento_total:.2f}")
```

***Identificación del cliente con mayor rendimiento:***

En el análisis financiero es fundamental evaluar el rendimiento generado por cada cliente a partir de sus inversiones. Determinar cuál cliente obtiene el mayor rendimiento permite a la empresa identificar perfiles más rentables, orientar mejor las estrategias comerciales y optimizar la asignación de recursos. Este proceso no solo facilita una visión más precisa del comportamiento de la cartera, sino que también apoya la toma de decisiones enfocadas en maximizar la rentabilidad y fortalecer la relación con los clientes más valiosos.

```python
# Cliente con mayor rendimiento absoluto
mayor_cliente = ""
mayor_rendimiento = 0
for cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento"]) / 100
    if rendimiento_cliente > mayor_rendimiento:
        mayor_rendimiento = rendimiento_cliente
        mayor_cliente = cliente

print(f"El cliente con mayor rendimiento absoluto es {mayor_cliente} con ${mayor_rendimiento:.2f}")
print("-"*50)
```

Ana Gómez es la cliente más rentable, ya que genera el mayor rendimiento absoluto de $1,350.00

***Parte 4: Modificaciones del Portafolio***

Carlos Lopez ha decidido aumentar su inversión a 25,000 dólares.

```python
# 1. Actualizar monto de Carlos Lopez
portafolio["Carlos Lopez"] = {"monto_invertido":  25000, "rendimiento": 4.0}
```

***Un nuevo cliente decide invertir***

Un nuevo cliente, Laura Fernandez, ha decidido invertir 22,000 dólares con un rendimiento del 6.8%.

```python
# 2. Añadir nuevo cliente
portafolio["Laura Fernandez"] = {"monto_invertido": 22000, "rendimiento": 6.8}
```

***Remover a un cliente de la inversión***

Ana Gomez ha decidido retirar su inversión.

```python
# 3. Remover a Ana Gomez
del portafolio["Ana Gomez"]
```

***Portalio actualizado***

```python
print("Parte 4 - Portafolio actualizado:")
for cliente, datos in portafolio.items():
    print(f"{cliente}: {datos}")
print("-"*100)
```
